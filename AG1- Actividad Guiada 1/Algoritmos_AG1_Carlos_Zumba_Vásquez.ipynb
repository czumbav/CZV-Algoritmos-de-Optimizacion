### Problema de los Puntos Más Cercanos

El problema de **los puntos más cercanos** (Closest Pair of Points Problem) es un problema de geometría computacional y diseño de algoritmos cuyo objetivo es determinar, dentro de un conjunto finito de puntos, cuáles son los dos que presentan la **menor distancia entre sí**.

Dado un conjunto de $n$ puntos:

$$P = \{p_1, p_2, \dots, p_n\}$$

se busca encontrar el par de puntos distintos:

$$(p_i, p_j), \quad i \neq j$$

que minimice una función de distancia $d$. Matemáticamente, el problema puede expresarse como:

$$(p_i^*, p_j^*) = \arg\min_{i \neq j} d(p_i, p_j)$$

Es decir, no se busca únicamente conocer la distancia mínima, sino identificar **qué dos puntos producen dicha distancia**.

En **una dimensión**, cada punto corresponde simplemente a un número real:

$$P = \{x_1, x_2, \dots, x_n\}$$

y la distancia entre dos elementos se define mediante el valor absoluto:

$$d(x_i, x_j) = |x_i - x_j|$$

Por ejemplo, si:

$$P = \{15, 18, 24, 55, 11\}$$

las distancias de algunos pares serían:

### Primer intento de resolución: Fuerza bruta 1D

El enfoque más inmediato para resolver el problema es mediante **fuerza bruta**. En donde, este procedimiento compara cada punto con todos los puntos posteriores y mantiene la menor distancia encontrada. Como el orden de los elementos del par no es relevante, $(x_i, x_j)$ y $(x_j, x_i)$ representan la misma solución. Por ello, el número total de pares que deben evaluarse es una combinación:

$$\binom{n}{2} = \frac{n(n - 1)}{2}$$

Por ejemplo, si tomamos los datos de la sección anterior determinados con 10 puntos:

$$\binom{5}{2} = \frac{5(4)}{2} = 10$$

Por tanto, fuerza bruta necesitó evaluar **10 pares**. En términos asintóticos:

$$\frac{n(n - 1)}{2} = \frac{n^2 - n}{2}$$

#### Implementación en Python: Puntos Más Cercanos en 1D mediante Fuerza Bruta
El siguiente código resuelve el problema de los puntos más cercanos en 1D aplicando la estrategia de Fuerza Bruta. Esta función encuentra los dos valores más cercanos comparando todos los pares posibles.

```python
# Definir función
def puntos_mas_cercanos_1d_fuerza_bruta(lista):
    if len(lista) < 2:
        raise ValueError("Se necesitan al menos dos puntos.")

    # distancia_minima = float("inf")
    distancia_minima = float("inf")
    # Variable para guardar los puntos con menor distancia
    mejor_par = None

    for i in range(len(lista)):
        for j in range(i + 1, len(lista)):
            # Calcular la distancia
            distancia = abs(lista[i] - lista[j])

            # Comparar con la mejor distancia
            if distancia < distancia_minima:
                distancia_minima = distancia
                mejor_par = (lista[i], lista[j])

    return mejor_par, distancia_minima
```

#### Implementación en Python: Puntos Más Cercanos en 1D mediante Fuerza Bruta (Mostrando todos los pares que tengan la misma distancia mínima)
El ajuste de la función devuelve todos los pares que tengan la misma distancia mínima, el cambio principal es sustituir `mejor_par` por una lista de mejores pares.

```python
# Definir función
def puntos_mas_cercanos_1d_fuerza_bruta_mejores_pares(lista):
    if len(lista) < 2:
        raise ValueError("Se necesitan al menos dos puntos.")

    distancia_minima = float("inf")
    # Lista para guardar todos los pares con distancia mínima
    mejores_pares = []

    for i in range(len(lista)):
        for j in range(i + 1, len(lista)):
            # Calcular la distancia
            distancia = abs(lista[i] - lista[j])

            # Comparar con la mejor distancia
            if distancia < distancia_minima:
                distancia_minima = distancia
                mejores_pares = [(lista[i], lista[j])]

            # Si se encuentra otra pareja de puntos con la misma distancia mínima, la agregamos a la lista
            elif distancia == distancia_minima:
                mejores_pares.append((lista[i], lista[j]))

    return mejores_pares, distancia_minima
```

#### Cálculo de la complejidad

Para el caso 1D, la complejidad depende del enfoque, para lo cuál se debe considerar dos estrategias: 
* La fuerza bruta sobre datos no ordenados y
* El algoritmo mejorado que primero ordena.

#### Fuerza bruta con datos no ordenados
En fuerza bruta no aprovechamos ninguna estructura especial de los datos. El objetivo es simplemente comparar cada punto con todos los demás puntos que todavía no han sido comparados. 

Tal como se lo realizo en el código de Python de la sección anterior en donde trabaja con **datos No ordenados**. Al trabajar con el número total de pares que se evaluarón a tráves de una combinación:

$$\binom{n}{2} = \frac{n(n - 1)}{2} = \frac{n^2 - n}{2}$$

y al considerar únicamente el término dominante, en análisis Big-O se conserva el término dominante:

$$\mathcal{O}(n^2)$$

Porque cuando $n$ crece, el término $n^2$ domina sobre $n$.

#### ¿Se puede mejorar?
#### Fuerza bruta con datos ordenados

La solución de fuerza bruta con datos no ordenados es una solución correcta y sencilla, pero pierde eficiencia conforme crece el número de elementos. En este sentido, se puede obtener una mejora del **algoritmo de fuerza bruta**, al considerar que en una dimensión se puede mejorar aprovechando una propiedad fundamental: si los puntos están ordenados, el par de puntos más cercano necesariamente debe encontrarse entre dos posiciones consecutivas.

Esta mejora establece que primero se necesita ordenar para resolverlo más eficientemente.

Retomando los datos del ejemplo inicial

$$P = \{15, 18, 24, 55, 11\}$$

Después de ordenar:

$$P' = \{11, 15, 18, 24, 55\}$$

ya no es necesario evaluar todos los pares. Basta calcular las diferencias entre elementos consecutivos:

$$15 - 11 = 4$$
$$18 - 15 = 3$$
$$24 - 18 = 6$$
$$55 - 24 = 31$$

La menor distancia vuelve a ser:

$$3$$

correspondiente a:

$$(15, 18)$$


El ordenamiento tiene una complejidad típica:

$$O(n \log n)$$

y el recorrido posterior:

$$O(n)$$

por lo que la complejidad total queda determinada por la operación dominante:

$$O(n \log n) + O(n) = O(n \log n)$$

La importancia académica del problema está precisamente en esta transición. Se parte de una solución intuitiva de fuerza bruta:

$$O(n^2)$$

y posteriormente se estudia cómo aprovechar la **estructura del problema** para reducir el espacio de comparaciones:

$$O(n \log n)$$

##### Implementación en Python: Puntos Más Cercanos en 1D mediante Fuerza Bruta (Mejora)
La mejora del algoritmo de fuerza bruta se basa en encontrar los dos puntos más cercanos en una dimensión mediante ordenamiento y luego realizar una comparación de puntos consecutivos (vecinos).

```python
# Definir función
def puntos_mas_cercanos_1d_fuerza_bruta_ordenado(lista):
    if len(lista) < 2:
        raise ValueError("Se necesitan al menos dos puntos.")

    # Ordenar los puntos de menor a mayor
    lista_ordenada = sorted(lista)
    # Inicializar la distancia mínima con infinito
    distancia_minima = float("inf")
    # Lista para guardar todos los pares con menor distancia
    mejores_pares = []

    # Comparar únicamente puntos consecutivos
    for i in range(len(lista_ordenada) - 1):
        # Calcular distancia entre el punto actual y el siguiente
        distancia = abs(lista_ordenada[i + 1] - lista_ordenada[i])

        # Comparar con la mejor distancia encontrada
        if distancia < distancia_minima:
            distancia_minima = distancia
            mejores_pares  = [lista_ordenada[i], lista_ordenada[i + 1]]

        # Si encontramos otra pareja con la misma distancia mínima,la agregamos a la lista
        elif distancia == distancia_minima:
            mejores_pares.append((lista_ordenada[i], lista_ordenada[i + 1]))

    return mejores_pares , distancia_minima
```

### Segundo intento. Aplicar Divide y Vencerás

Divide y Vencerás plantea otra estrategia, en lugar de resolver directamente el problema completo, dividirlo en problemas más pequeños, resolverlos y combinar sus soluciones. En este sentido, para el problema de los puntos más cercanos en 1D, la técnica de Divide y Vencerás consiste en ordenar primero los puntos, dividir el conjunto en dos mitades, resolver recursivamente el problema en cada mitad y, finalmente, comprobar si existe un par más cercano formado por los puntos situados en la frontera de ambas particiones.

Una vez que los puntos están ordenados, dividimos el problema en dos mitades; en donde la recurrencia de la fase recursiva en 1D puede expresarse como:

$$T(n) = 2T \left( \frac{n}{2} \right) + O(1)$$

¿Por qué?

* **$2T(n/2)$:** resolvemos dos subproblemas de tamaño $n/2$.
* **$O(1)$:** al combinar, en 1D basta comparar los dos puntos que quedan en la frontera de las mitades, si la implementación conserva correctamente los extremos.

#### Implementación en Python: Puntos Más Cercanos en 1D mediante Divide y Vencerás
El siguiente código resuelve el problema de los puntos más cercanos en 1D aplicando la estrategia Divide y Vencerás usando datos no ordenados.

```python
# Definir función
def puntos_mas_cercanos_1d_divide_venceras(lista):
    """
    Encuentra los dos puntos más cercanos en 1D utilizando Divide y Vencerás sobre datos no ordenados.
    Divide: Divide la lista en dos mitades.
    Vence: Resuelve recursivamente cada mitad.
    Combina: Compara los puntos de una mitad contra los de la otra.
    Complejidad: O(n²)
    """

    if len(lista) < 2:
        return [], float("inf")

    # ----------------------------------------
    # CASO BASE
    # ----------------------------------------
    if len(lista) == 2:
        par = (lista[0], lista[1])
        distancia = abs(lista[0] - lista[1])

        return [par], distancia

    # ----------------------------------------
    # DIVIDIR
    # ----------------------------------------
    medio = len(lista) // 2
    izquierda = lista[:medio]
    derecha = lista[medio:]

    # ----------------------------------------
    # VENCER
    # ----------------------------------------
    pares_izq, distancia_izq = puntos_mas_cercanos_1d_divide_venceras(izquierda)
    pares_der, distancia_der = puntos_mas_cercanos_1d_divide_venceras(derecha)

    # Seleccionar inicialmente la mejor solución de izquierda y derecha
    if distancia_izq < distancia_der:    
        distancia_minima = distancia_izq
        mejores_pares = pares_izq.copy()
    elif distancia_der < distancia_izq:
        distancia_minima = distancia_der
        mejores_pares = pares_der.copy()    
    else:
        # Si ambas mitades tienen la misma distancia mínima, conservamos los pares de ambos lados
        distancia_minima = distancia_izq
        mejores_pares = (pares_izq.copy() + pares_der.copy())

    # ----------------------------------------
    # COMBINAR: Como no existe ordenamiento, debemos comprobar todos los pares que cruzan ambas mitades
    # ----------------------------------------
    for punto_izq in izquierda:
        for punto_der in derecha:
            distancia = abs(punto_izq - punto_der)

            if distancia < distancia_minima:
                distancia_minima = distancia
                mejores_pares = [(punto_izq, punto_der)]
            # Si encontramos otro par con la misma distancia mínima, lo agregamos a la lista
            elif distancia == distancia_minima:
                mejores_pares.append((punto_izq, punto_der))

    return mejores_pares, distancia_minima
```

```python
# Aplicar solución del algoritmo
par, distancia = puntos_mas_cercanos_1d_divide_venceras(LISTA_1D)

print("Puntos más cercanos:", par)
print("Distancia mínima:", distancia)
```

##### Calculo de la complejidad

Esta fase tiene complejidad:

$$O(n)$$

si los datos ya están ordenados.

Sin embargo, cuando la entrada está desordenada es necesario realizar previamente un ordenamiento:

$$O(n \log n)$$

Por tanto, la complejidad total es:

$$O(n \log n) + O(n)$$

y, al conservar el término dominante:

$$O(n \log n)$$

En consecuencia, Divide y Vencerás mejora el enfoque de fuerza bruta, que requiere:

$$O(n^2)$$

pasando a:

$$O(n \log n)$$

#### Implementación en Python: Puntos Más Cercanos en 1D mediante Divide y Vencerás (Mejora)
En 1D, además, el problema puede simplificarse todavía más: una vez ordenados los puntos, basta comparar elementos consecutivos, por lo que no es estrictamente necesario implementar toda la recursividad de Divide y Vencerás. Si los datos ya vienen ordenados, la solución puede obtenerse directamente en:

```python
# Definir función
def puntos_mas_cercanos_1d_divide_venceras_mejora(lista, inicio=None, fin=None):
    """
    Encuentra los dos puntos más cercanos en 1D mediante Divide y Vencerás optimizado. La lista se ordena una sola vez en la primera llamada.
    Divide: Divide el intervalo en dos mitades.
    Vence: Resuelve recursivamente ambas mitades.
    Combina: Compara únicamente los puntos de la frontera.
    Complejidad:
        Ordenamiento:      O(n log n)
        Divide y Vencerás: O(n)
        Total:             O(n log n)
    """
    # ----------------------------------------
    # PRIMERA LLAMADA
    # ----------------------------------------
    if inicio is None:
        if len(lista) < 2:
            raise ValueError("Se necesitan al menos dos puntos.")

        # Ordenar una sola vez
        lista = sorted(lista)
        inicio = 0
        fin = len(lista) - 1

    # ----------------------------------------
    # CASO BASE 1: # Un solo punto
    # ----------------------------------------
    if inicio == fin:
        return [], float("inf")

    # ----------------------------------------
    # CASO BASE 2: # Dos puntos
    # ----------------------------------------
    if fin - inicio == 1:
        par = (lista[inicio], lista[fin])
        distancia = (lista[fin] - lista[inicio])

        return [par], distancia

    # ----------------------------------------
    # DIVIDIR
    # ----------------------------------------
    medio = (inicio + fin) // 2
    # print("Punto Medio " + str(medio))
    
    # ----------------------------------------
    # VENCER
    # ----------------------------------------
    pares_izq, distancia_izq = (puntos_mas_cercanos_1d_divide_venceras_mejora(lista, inicio, medio))
    pares_der, distancia_der = (puntos_mas_cercanos_1d_divide_venceras_mejora(lista, medio + 1, fin))
    # print("Comprobando puntos de la izquierda" + str(pares_izq) + " y " + str(distancia_izq))
    # print("Comprobando puntos de la perecha" + str(pares_der) + " y " + str(distancia_der))

    # ----------------------------------------
    # COMBINAR: Solo comparar la frontera
    # ----------------------------------------
    par_frontera = (lista[medio], lista[medio + 1])
    distancia_frontera = (lista[medio + 1]- lista[medio])

    # ----------------------------------------
    # SELECCIONAR LA DISTANCIA MÍNIMA
    # ----------------------------------------
    distancia_minima = min(distancia_izq, distancia_der, distancia_frontera)

    # Lista para guardar todos los pares que tengan la distancia mínima
    mejores_pares = []

    # ----------------------------------------
    # AGREGAR PARES DE LA IZQUIERDA
    # ----------------------------------------
    if distancia_izq == distancia_minima:
        mejores_pares.extend(pares_izq)

    # ----------------------------------------
    # AGREGAR PARES DE LA DERECHA
    # ----------------------------------------
    if distancia_der == distancia_minima:
        mejores_pares.extend(pares_der)

    # ----------------------------------------
    # AGREGAR PAR DE LA FRONTERA
    # ----------------------------------------
    if distancia_frontera == distancia_minima:
        mejores_pares.append(par_frontera)

    return mejores_pares, distancia_minima
```

```python
# Aplicar solución del algoritmo
par, distancia = puntos_mas_cercanos_1d_divide_venceras_mejora(LISTA_1D)

print("Puntos más cercanos:", par)
print("Distancia mínima:", distancia)
```

### Extender el algoritmo a 2D: [(1122, 6175), (135, 4076), (7296, 2741)..

Para extender el algoritmo a **2D**, ya no se puede usar simplemente `abs(x1 - x2)`, porque cada punto tiene dos coordenadas:

$$P_i = (x_i, y_i)$$

La distancia entre dos puntos se calcula mediante la **distancia euclidiana**:

$$d(P_i, P_j) = \sqrt{(x_i - x_j)^2 + (y_i - y_j)^2}$$

Además, en **fuerza bruta 2D no basta con ordenar y comparar puntos consecutivos**. Debemos comparar **todos los pares posibles**, porque dos puntos cercanos geométricamente no necesariamente quedan consecutivos al ordenar solo por $x$ o por $y$.

#### Implementación en Python: Puntos Más Cercanos en 2D mediante Fuerza Bruta

```python
# Definir función
def puntos_mas_cercanos_2d_fuerza_bruta(lista):
    """
    Encuentra los dos puntos más cercanos en 2D mediante Fuerza Bruta.
    Cada punto tiene la forma: (x, y)
    Complejidad: O(n²)
    """

    if len(lista) < 2:
        raise ValueError("Se necesitan al menos dos puntos.")

    # Inicializar la distancia mínima con infinito
    distancia_minima = float("inf")

    # Lista para guardar todos los pares que tengan la menor distancia
    mejores_pares = []

    # Comparar todos los pares posibles
    for i in range(len(lista)):
        for j in range(i + 1, len(lista)):
            # Obtener las coordenadas de cada punto
            x1, y1 = lista[i]
            x2, y2 = lista[j]

            # Calcular la distancia euclidiana
            distancia = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

            # Comparar con la mejor distancia encontrada
            if distancia < distancia_minima:
                distancia_minima = distancia
                mejores_pares = [(lista[i], lista[j])]

            # Si encontramos otra pareja con. la misma distancia mínima
            elif distancia == distancia_minima:
                # Agregar el nuevo par
                mejores_pares.append((lista[i], lista[j]))

    return mejores_pares, distancia_minima
```

#### Implementación en Python: Puntos Más Cercanos en 2D mediante Divide y Vencerás
Para extender correctamente **Divide y Vencerás de 1D a 2D**, no basta con comparar únicamente los dos puntos de la frontera. Esa estrategia funciona en 1D, pero **no en 2D**, porque un punto de la mitad izquierda puede estar muy cerca de varios puntos de la mitad derecha.

En 2D debemos trabajar con una **franja (*strip*) alrededor de la línea divisoria** y comprobar los puntos candidatos dentro de ella. Así conservamos la solución exacta y una complejidad:

$$O(n \log n)$$

```python
def puntos_mas_cercanos_2d_divide_venceras(lista):
    """
    Encuentra los dos puntos más cercanos en 2D mediante Divide y Vencerás optimizado.
    Cada punto tiene la forma: (x, y)
    Estrategia:
        1. Ordenar los puntos por X y por Y.
        2. Dividir el conjunto en dos mitades.
        3. Resolver recursivamente cada mitad.
        4. Crear una franja alrededor de la división.
        5. Comprobar los candidatos de la franja.
        6. Seleccionar la menor distancia.
    Complejidad:
        Para encontrar la distancia mínima:
            O(n log n)
        Para devolver todos los pares empatados:
            depende del número de pares encontrados.
        
        En el peor caso, si existen O(n²) pares mínimos, la complejidad total puede llegar a O(n²).
    """

    if len(lista) < 2:
        raise ValueError("Se necesitan al menos dos puntos.")

    # -------------------------------------------------
    # Agregar un identificador a cada punto, esto permite manejar correctamente puntos repetidos
    # -------------------------------------------------
    puntos = [
        (x, y, i)
        for i, (x, y) in enumerate(lista)
    ]

    # Ordenar una sola vez por coordenada X
    puntos_x = sorted(puntos, key=lambda punto: (punto[0], punto[1], punto[2]))

    # Ordenar una sola vez por coordenada Y
    puntos_y = sorted(puntos, key=lambda punto: (punto[1], punto[0], punto[2]))

    # -------------------------------------------------
    # Función para calcular distancia al cuadrado, evita calcular sqrt en todas las comparaciones
    # -------------------------------------------------
    def distancia_cuadrado(p1, p2):
        return ((p1[0] - p2[0]) ** 2 + (p1[1] - p2[1]) ** 2)

    # -------------------------------------------------
    # Función auxiliar: Eliminar pares repetidos
    # -------------------------------------------------
    def eliminar_pares_repetidos(pares):
        """
        Evita que el mismo par aparezca varias veces debido a que puede encontrarse tanto en una llamada recursiva como nuevamente en la franja.
        Los identificadores permiten saber si se trata realmente del mismo par de puntos.
        """
        pares_unicos = []
        ids_encontrados = set()

        for p1, p2 in pares:
            # Crear identificador independiente del orden
            clave = tuple(sorted((p1[2], p2[2])))

            if clave not in ids_encontrados:
                ids_encontrados.add(clave)
                pares_unicos.append((p1, p2))

        return pares_unicos
    
    # -------------------------------------------------
    # Función Recursiva
    # -------------------------------------------------
    def dividir_vencer(px, py):
        n = len(px)
        # ---------------------------------------------
        # CASO BASE: Un solo punto
        # ---------------------------------------------
        if n < 2:
            return [], float("inf")
            # ---------------------------------------------
        # CASO BASE: Para 2 o 3 puntos aplicar fuerza bruta
        # ---------------------------------------------
        if n <= 3:
            distancia_minima = float("inf")
            # Ahora tenemos una lista de mejores pares
            mejores_pares = []

            for i in range(n):
                for j in range(i + 1, n):
                    distancia = distancia_cuadrado(px[i], px[j])

                    # Encontramos una distincia menor
                    if distancia < distancia_minima:
                        distancia_minima = distancia
                        mejores_pares = [(px[i], px[j])]
                    # ---------------------------------
                    # Encontramos otro par con exactamente la misma distancia
                    # ---------------------------------
                    elif distancia == distancia_minima:
                        mejores_pares.append((px[i], px[j]))

            return mejores_pares, distancia_minima

        # ---------------------------------------------
        # DIVIDIR
        # ---------------------------------------------
        medio = n // 2
        izquierda_x = px[:medio]
        derecha_x = px[medio:]

        # Coordenada X donde se realiza la división
        x_medio = px[medio][0]

        # Identificadores de los puntos de la izquierda
        ids_izquierda = {
            punto[2]
            for punto in izquierda_x
        }

        # Mantener también las dos mitades ordenadas por Y
        izquierda_y = [
            punto
            for punto in py
            if punto[2] in ids_izquierda
        ]

        derecha_y = [
            punto
            for punto in py
            if punto[2] not in ids_izquierda
        ]

        # ---------------------------------------------
        # VENCER: Resolver recursivamente izquierda y derecha
        # ---------------------------------------------
        pares_izq, distancia_izq = dividir_vencer(izquierda_x, izquierda_y)
        pares_der, distancia_der = dividir_vencer(derecha_x, derecha_y)

        # Seleccionar inicialmente la mejor solución
        if distancia_izq < distancia_der:
            distancia_minima = distancia_izq
            mejores_pares = pares_izq.copy()
        elif distancia_der < distancia_izq:
            distancia_minima = distancia_der
            mejores_pares = pares_der.copy()
        else:
            # -----------------------------------------
            # Si izquierda y derecha tienen exactamente. la misma distancia mínima, conservar los pares de AMBOS lados.
            # -----------------------------------------
            distancia_minima = distancia_izq
            mejores_pares = (pares_izq.copy() + pares_der.copy())

        # ---------------------------------------------
        # COMBINAR: Crear una franja alrededor de la línea divisoria utilizando la distancia mínima encontrada hasta este momento.
        # ---------------------------------------------
        franja = [
            punto
            for punto in py
            if (punto[0] - x_medio) ** 2
            <= distancia_minima
        ]

        # ---------------------------------------------
        # Examinar candidatos dentro de la franja, como la franja está ordenada por Y, solo es necesario revisar unos pocos vecinos.
        # ---------------------------------------------
        for i in range(len(franja)):
            for j in range(i + 1, len(franja)):
                # Diferencia vertical al cuadrado
                diferencia_y = (franja[j][1] - franja[i][1]) ** 2

                # -------------------------------------
                # Como la franja está ordenada por Y, si ya SUPERAMOS la distancia mínima,
                # los siguientes puntos tampoco podrán mejorar ni empatar.
                # -------------------------------------
                if diferencia_y > distancia_minima:
                    break

                distancia = distancia_cuadrado(franja[i], franja[j])

                # -------------------------------------
                # Encontramos una distancia MENOR
                # -------------------------------------
                if distancia < distancia_minima:
                    distancia_minima = distancia
                    # Eliminar todas las soluciones anteriores
                    mejores_pares = [(franja[i], franja[j])]

                # -------------------------------------
                # Encontramos otra pareja con la MISMA distancia mínima
                # -------------------------------------
                elif distancia == distancia_minima:
                    mejores_pares.append((franja[i], franja[j]))

        # ---------------------------------------------
        # Eliminar pares que pudieron encontrarse tanto recursivamente como en la franja
        # ---------------------------------------------
        mejores_pares = eliminar_pares_repetidos(mejores_pares)

        return (mejores_pares, distancia_minima)

    # -------------------------------------------------
    # INICIAR DIVIDE Y VENCERÁS
    # -------------------------------------------------
    mejores_pares, distancia_cuadrado_minima = (dividir_vencer(puntos_x, puntos_y))

    # -------------------------------------------------
    # Recuperar solamente las coordenadas (x, y) de TODOS los mejores pares
    # -------------------------------------------------
    mejores_pares_coordenadas = []

    for p1, p2 in mejores_pares:
        punto_1 = (p1[0], p1[1])
        punto_2 = (p2[0], p2[1])

        mejores_pares_coordenadas.append((punto_1, punto_2))

    # -------------------------------------------------
    # Calcular la distancia real solamente al final
    # -------------------------------------------------
    distancia_minima = (distancia_cuadrado_minima ** 0.5)

    return (mejores_pares_coordenadas, distancia_minima)
```

```python
par, distancia = puntos_mas_cercanos_2d_divide_venceras(LISTA_2D)

print("\nPuntos más cercanos:", par)
print("Distancia mínima:", distancia)
```

### Extender el algoritmo a 3D

Para extender la solución del algoritmo por **Fuerza Bruta** de 2D a **3D**, solo se debe cambiar la representación de cada punto y la fórmula de distancia.

En 3D cada punto tiene la forma:

$$(x, y, z)$$

y la distancia euclidiana entre dos puntos es:

$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2 + (z_2 - z_1)^2}$$

#### Implementación en Python: Puntos Más Cercanos en 3D mediante Fuerza Bruta

```python
# Definir función
def puntos_mas_cercanos_3d_fuerza_bruta(lista):
    """
    Encuentra los dos puntos más cercanos en 3D mediante Fuerza Bruta.
    Cada punto tiene la forma: (x, y, z)
    Complejidad: O(n²)
    """

    if len(lista) < 2:
        raise ValueError("Se necesitan al menos dos puntos.")

    # Inicializar la distancia mínima con infinito
    distancia_minima = float("inf")

    # Variable para guardar los puntos con menor distancia
    mejores_pares = []

    # Comparar todos los pares posibles
    for i in range(len(lista)):
        for j in range(i + 1, len(lista)):
            # Obtener las coordenadas de cada punto
            x1, y1, z1 = lista[i]
            x2, y2, z2 = lista[j]

            # Calcular la distancia euclidiana en 3D
            distancia = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2 + (z2 - z1) ** 2)

            # Comparar con la mejor distancia encontrada
            if distancia < distancia_minima:
                # Actualizar la nueva distancia mínima
                distancia_minima = distancia
                # Eliminar los pares anteriores y guardar solamente el nuevo mejor par
                mejores_pares = [(lista[i], lista[j])]
                
            # Si encontramos otro par con la misma distancia mínima
            elif distancia == distancia_minima:
                # Agregar el nuevo par a la lista
                mejores_pares.append((lista[i], lista[j]))

    return mejores_pares, distancia_minima
```

#### Implementación en Python: Puntos Más Cercanos en 3D mediante Divide y Vencerás
Para extender correctamente el algoritmo Divide y Vencerás de 2D a **3D**, hay una diferencia importante a considerar: en 2D se usa la propiedad de revisar hasta 7 vecinos en la franja; **ese "7" no puede trasladarse directamente a 3D**. En 3D conviene mantener la división por $x$ y, durante la combinación, usar una pequeña estructura espacial tipo rejilla para localizar candidatos cercanos.

Cada punto ahora tiene la forma:

$$P = (x, y, z)$$

y la distancia euclidiana es:

$$d(P_1, P_2) = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2 + (z_2 - z_1)^2}$$

```python
# Definir función
def puntos_mas_cercanos_3d_divide_venceras(lista):
    """
    Encuentra los puntos más cercanos en 3D mediante
    Divide y Vencerás.

    Cada punto tiene la forma:
        (x, y, z)

    Estrategia:
        1. Ordenar los puntos por X y por Y.
        2. Dividir el conjunto en dos mitades.
        3. Resolver recursivamente cada mitad.
        4. Crear una franja alrededor del plano de división.
        5. Comprobar los candidatos de la franja.
        6. Seleccionar todos los pares con distancia mínima.

    Complejidad:
        Para encontrar la distancia mínima:
            O(n log n) en el enfoque Divide y Vencerás.

        Para devolver todos los pares empatados:
            depende del número de pares encontrados.

        En el peor caso, si existen O(n²) pares mínimos,
        la complejidad total puede llegar a O(n²).
    """

    if len(lista) < 2:
        raise ValueError(
            "Se necesitan al menos dos puntos."
        )

    # -------------------------------------------------
    # Agregar un identificador a cada punto.
    # El punto original:
    #     (x, y, z)
    # pasa internamente a:
    #     (x, y, z, id)
    # Esto permite manejar correctamente puntos repetidos.
    # -------------------------------------------------
    puntos = [(x, y, z, i) for i, (x, y, z) in enumerate(lista)]

    # -------------------------------------------------
    # Ordenar una sola vez por coordenada X
    # -------------------------------------------------
    puntos_x = sorted(puntos, key=lambda punto: (punto[0], punto[1], punto[2]))

    # -------------------------------------------------
    # Ordenar una sola vez por coordenada Y
    # -------------------------------------------------
    puntos_y = sorted(puntos, key=lambda punto: (punto[1], punto[0], punto[2]))

    # -------------------------------------------------
    # Función para calcular distancia al cuadrado. Evita calcular sqrt en todas las comparaciones.
    # -------------------------------------------------
    def distancia_cuadrado(p1, p2):
        return ((p1[0] - p2[0]) ** 2 + (p1[1] - p2[1]) ** 2 + (p1[2] - p2[2]) ** 2)

    # -------------------------------------------------
    # Función auxiliar: Eliminar pares repetidos
    # -------------------------------------------------
    def eliminar_pares_repetidos(pares):
        """
        Evita que el mismo par aparezca varias veces. Como cada punto tiene un identificador único, podemos distinguir puntos con las mismas coordenadas.
        """
        pares_unicos = []
        ids_encontrados = set()

        for p1, p2 in pares:
            # En 3D el identificador está en la posición 3.
            clave = tuple(sorted((p1[3], p2[3])))

            if clave not in ids_encontrados:
                ids_encontrados.add(clave)
                pares_unicos.append((p1, p2))

        return pares_unicos

    # -------------------------------------------------
    # FUNCIÓN RECURSIVA
    # -------------------------------------------------
    def dividir_vencer(px, py):
        n = len(px)
        # ---------------------------------------------
        # CASO BASE: Un solo punto
        # ---------------------------------------------
        if n < 2:
            return [], float("inf")

        # ---------------------------------------------
        # CASO BASE: Para 2 o 3 puntos aplicar Fuerza Bruta
        # ---------------------------------------------
        if n <= 3:
            distancia_minima = float("inf")
            mejores_pares = []
            for i in range(n):
                for j in range(i + 1, n):
                    distancia = distancia_cuadrado(px[i],px[j])

                    # ---------------------------------
                    # Encontramos una distancia menor
                    # ---------------------------------
                    if distancia < distancia_minima:
                        distancia_minima = distancia
                        mejores_pares = [(px[i], px[j])]

                    # ---------------------------------
                    # Encontramos otro par con exactamente la misma distancia
                    # ---------------------------------
                    elif distancia == distancia_minima:
                        mejores_pares.append((px[i], px[j]))

            return (mejores_pares, distancia_minima)

        # ---------------------------------------------
        # DIVIDIR
        # ---------------------------------------------
        medio = n // 2
        izquierda_x = px[:medio]
        derecha_x = px[medio:]
        
        # Plano X donde se realiza la división
        x_medio = px[medio][0]

        # ---------------------------------------------
        # Identificadores de los puntos pertenecientes a la izquierda. En 3D el id está en punto[3].
        # ---------------------------------------------
        ids_izquierda = {
            punto[3]
            for punto in izquierda_x
        }

        # ---------------------------------------------
        # Mantener las dos mitades ordenadas por Y
        # ---------------------------------------------
        izquierda_y = [
            punto
            for punto in py
            if punto[3] in ids_izquierda
        ]

        derecha_y = [
            punto
            for punto in py
            if punto[3] not in ids_izquierda
        ]

        # ---------------------------------------------
        # VENCER: Resolver recursivamente izquierda y derecha
        # ---------------------------------------------
        pares_izq, distancia_izq = dividir_vencer(izquierda_x, izquierda_y)
        pares_der, distancia_der = dividir_vencer(derecha_x, derecha_y)

        # ---------------------------------------------
        # Seleccionar inicialmente las mejores soluciones
        # ---------------------------------------------
        if distancia_izq < distancia_der:
            distancia_minima = distancia_izq
            mejores_pares = pares_izq.copy()
        elif distancia_der < distancia_izq:
            distancia_minima = distancia_der
            mejores_pares = pares_der.copy()
        else:
            # -----------------------------------------
            # Si izquierda y derecha tienen la misma distancia mínima, conservar los pares de ambos lados.
            # -----------------------------------------
            distancia_minima = distancia_izq
            mejores_pares = (pares_izq.copy() + pares_der.copy())

        # ---------------------------------------------
        # COMBINAR: Crear una franja alrededor del plano X, distancia_minima está al cuadrado.
        # ---------------------------------------------
        franja = [
            punto
            for punto in py
            if (
                punto[0] - x_medio
            ) ** 2 <= distancia_minima
        ]

        # ---------------------------------------------
        # Examinar candidatos dentro de la franja. La franja está ordenada por Y.
        # ---------------------------------------------
        for i in range(len(franja)):
            for j in range(i + 1, len(franja)):
                # -------------------------------------
                # Diferencia en Y al cuadrado
                # -------------------------------------
                diferencia_y = (franja[j][1] - franja[i][1]) ** 2

                # Como la franja está ordenada por Y, si la diferencia en Y supera la distancia mínima, ya no es necesario continuar.
                if diferencia_y > distancia_minima:
                    break

                # -------------------------------------
                # Diferencia en Z al cuadrado
                # -------------------------------------
                diferencia_z = (franja[j][2] - franja[i][2]) ** 2

                # Si solo la diferencia en Z ya supera la distancia mínima, este par no puede ser candidato.
                if diferencia_z > distancia_minima:
                    continue

                # -------------------------------------
                # Calcular distancia 3D
                # -------------------------------------
                distancia = distancia_cuadrado(franja[i], franja[j])

                # -------------------------------------
                # Encontramos una distancia MENOR
                # -------------------------------------
                if distancia < distancia_minima:
                    distancia_minima = distancia
                    # Eliminar soluciones anteriores
                    mejores_pares = [(franja[i], franja[j])]

                # -------------------------------------
                # Encontramos otra pareja con la misma distancia mínima
                # -------------------------------------
                elif distancia == distancia_minima:
                    mejores_pares.append((franja[i], franja[j]))

        # ---------------------------------------------
        # Eliminar pares que pudieron encontrarse tanto recursivamente como en la franja
        # ---------------------------------------------
        mejores_pares = eliminar_pares_repetidos(mejores_pares)

        return (mejores_pares, distancia_minima)

    # -------------------------------------------------
    # INICIAR DIVIDE Y VENCERÁS
    # -------------------------------------------------
    mejores_pares, distancia_cuadrado_minima = (dividir_vencer(puntos_x, puntos_y))

    # -------------------------------------------------
    # Recuperar solamente las coordenadas originales:
    # -------------------------------------------------
    mejores_pares_coordenadas = []

    for p1, p2 in mejores_pares:
        punto_1 = (p1[0], p1[1], p1[2])
        punto_2 = (p2[0], p2[1], p2[2])
        mejores_pares_coordenadas.append((punto_1, punto_2))

    # -------------------------------------------------
    # Calcular la distancia real solamente al final
    # -------------------------------------------------
    distancia_minima = (distancia_cuadrado_minima ** 0.5)

    return (mejores_pares_coordenadas, distancia_minima)
```
